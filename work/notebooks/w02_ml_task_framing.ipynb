## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision@10

I will use Precision@10 to evaluate the usefulness of the top 10 webpages recommended for content review. The dataset's trend_direction will be used as a proxy for identifying webpages with declining performance.

My initial target is at least 70% Precision@10, meaning at least 7 of the top 10 recommendations should show declining performance according to this proxy. This is a proposed target, not an achieved result. Human review will ultimately be needed to confirm whether these webpages genuinely need refreshing.

```python
# Section 3: Define the success metric using Hugging Face data

!pip install -q datasets huggingface_hub

import pandas as pd
from datasets import load_dataset

# Authenticate if the Hugging Face dataset is gated
from huggingface_hub import notebook_login
notebook_login()

# Load the actual FlyRank dataset
ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    streaming=True,
    split="train"
)

# Read a sample of records
records = list(ds.take(1000))
df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No records loaded from Hugging Face.")

# Check required columns (corrected names)
required_columns = ["content_hash_id"]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing required columns: {missing_columns}")

# Define the success metric
success_metric = "Precision@10"
top_k = 10
target_precision = 0.70

# Calculate the minimum relevant recommendations
minimum_relevant = int(top_k * target_precision)

print("\nSUCCESS METRIC")
print("-" * 50)
print("Dataset: FlyRank/internship-warehouse")
print("Configuration: fact_content_daily_performance")
print("Records inspected:", len(df))
print("Metric:", success_metric)
print("Top recommendations:", top_k)
print("Target precision:", target_precision * 100, "%")
print("Minimum relevant recommendations:", minimum_relevant)

print("\nA good result means at least 7 out of the top 10")
print("recommendations show declining performance according")
print("to the trend indicators available in the data.")
```

## 4. The unit of analysis, as a real dataframe

nit of Analysis: One Content Performance Record

Each row represents a daily content performance record for a webpage. The content_id identifies the content item, while the other columns describe its performance, such as impressions, clicks, sessions and engagement.

The model will analyse these records to identify content performance patterns and help prioritise webpages for human review.

```python
# Section 4: Unit of analysis as a real DataFrame

!pip install -q datasets huggingface_hub

import pandas as pd
from datasets import load_dataset
from huggingface_hub import notebook_login

# 1. Authenticate with Hugging Face
notebook_login()

# 2. Load the actual FlyRank dataset
ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    streaming=True,
    split="train"
)

# 3. Load a sample of records into a DataFrame
records = list(ds.take(1000))
df = pd.DataFrame(records)

if df.empty:
    raise ValueError("No records loaded from Hugging Face.")

# 4. Display dataset information
print("DATASET OVERVIEW")
print("-" * 50)

print("Dataset: FlyRank/internship-warehouse")
print("Configuration: fact_content_daily_performance")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

# 5. Display available columns
print("\nAVAILABLE COLUMNS:")
print(df.columns.tolist())

# 6. Check content identifier
if "content_id" in df.columns:
    print("\nUNIQUE CONTENT ITEMS:")
    print(df["content_id"].nunique())

    print("\nRecords per content item:")
    print(df["content_id"].value_counts().head())

else:
    print("\ncontent_id column not found. Inspect the available columns.")

# 7. Display the actual DataFrame
print("\nSAMPLE CONTENT PERFORMANCE RECORDS:")
display(df.head(10))

# 8. Display data types
print("\nDATA TYPES:")
display(df.dtypes.to_frame(name="Data Type"))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why Machine Learning Instead of Fixed Rules?

Content performance depends on several factors, including impressions, clicks, sessions, engagement rate, content age and recent performance trends. A fixed rule based on one threshold may miss important patterns and relationships between these features.

Machine Learning can analyse multiple features together and identify complex patterns in content performance. It can help rank webpages according to their likelihood of needing review, rather than relying on a single manually defined condition.

However, ML recommendations still require human validation and do not guarantee that refreshing content will improve performance.

```python
# Section 5: Why ML beats a fixed rule here

# 1. Define actual relevant features present in the dataset
features = [
    "gsc_impressions",
    "gsc_clicks",
    "gsc_avg_position",
    "sessions_organic",
    "ga4_pageviews"
]

# 2. Check which features are available
available_features = [
    col for col in features if col in df.columns
]

missing_features = [
    col for col in features if col not in df.columns
]

print("FEATURE AVAILABILITY")
print("-" * 50)
print("Available features:", available_features)
print("Missing features:", missing_features)

# 3. Display summary statistics
print("\nFEATURE SUMMARY:")
if available_features:
    display(df[available_features].describe())
else:
    print("No features found to describe.")

# 4. Demonstrate why multiple features matter
print("\nEXAMPLE: CONTENT PERFORMANCE PATTERNS")
print("-" * 50)

# Grouping by a categorical indicator to show pattern relationships
if "client_has_ga4" in df.columns and available_features:
    print("Mean features grouped by client_has_ga4:")
    display(
        df.groupby("client_has_ga4")[available_features]
          .mean(numeric_only=True)
    )
else:
    print("Grouping column or features are unavailable.")

# 5. Explain the limitation of a fixed rule
print("\nCONCLUSION")
print("-" * 50)
print("A single threshold cannot capture every content performance pattern.")
print("ML can analyse multiple features together and identify more complex patterns.")
print("Final recommendations should be validated by human reviewers.")
```
