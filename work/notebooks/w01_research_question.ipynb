## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

My work can identify patterns and relationships in the available search performance, engagement, and content-age data. It can provide directional insights and rank webpages that may deserve further review. These recommendations will support the content or SEO team in making informed decisions.

However, this analysis cannot prove that refreshing a webpage will directly improve its traffic or engagement. It cannot establish cause-and-effect relationships or predict Google's future ranking decisions. The scores are decision-support recommendations, not guaranteed outcomes, and human review will remain important.

```python
# Define the claims and limitations of this project

can_claim = [
    "Observed patterns in available data",
    "Directional insights",
    "Ranked recommendations for human review"
]

cannot_claim = [
    "Causal proof that refreshing content improves performance",
    "Prediction of Google's ranking decisions",
    "Guaranteed traffic or engagement improvement"
]

print("WHAT I CAN CLAIM:")
for item in can_claim:
    print("-", item)

print("\nWHAT I CANNOT CLAIM:")
for item in cannot_claim:
    print("-", item)
```
