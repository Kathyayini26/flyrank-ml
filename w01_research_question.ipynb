## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision my work will improve is which webpages should be prioritised for content review. The content or SEO team will use the ranked recommendations to decide whether a page needs refreshing, expanding, protecting, pruning, or monitoring. A wrong recommendation could waste the team's time and resources on low-priority pages while important pages that need attention are overlooked. Therefore, the scores will support human review rather than make automatic decisions.

```python

# Section 2: Decision, action and cost of a wrong recommendation
# Using the actual FlyRank Hugging Face dataset (with fallback to local repo CSV)

!pip install -q datasets

import os
import subprocess
import pandas as pd
from datasets import load_dataset

# Clone repository for fallback data
repo_url = "https://github.com/Kathyayini26/flyrank-ml.git"
repo_dir = "/content/flyrank-ml"

if not os.path.exists(repo_dir):
    try:
        subprocess.run(
            ["git", "clone", repo_url, repo_dir],
            check=True
        )
    except Exception as e:
        print("Repository cloning failed:", e)

# 1. Try loading the actual Hugging Face dataset
try:
    ds = load_dataset(
        "FlyRank/internship-warehouse",
        "fact_content_daily_performance",
        streaming=True,
        split="train"
    )

    records = list(ds.take(1000))
    df = pd.DataFrame(records)
    dataset_source = "Hugging Face (FlyRank/internship-warehouse)"

except Exception as e:
    print(f"Could not load gated HF dataset ({e}).")
    print("Falling back to cloned CSV repository data...\n")

    file_path = os.path.join(
        repo_dir, "data", "raw", "content_refresh_anonymized.csv"
    )

    if os.path.exists(file_path):
        df = pd.read_csv(file_path)
        dataset_source = f"Local Repository CSV ({file_path})"
    else:
        raise FileNotFoundError(
            "Hugging Face dataset is inaccessible and fallback CSV is not found."
        )

if df.empty:
    raise ValueError("No records were returned from the dataset.")

# 2. Inspect the actual dataset
print("DATASET OVERVIEW")
print("-" * 50)
print("Source:", dataset_source)
print("Records inspected:", len(df))
print("Number of columns:", len(df.columns))

print("\nAVAILABLE FEATURES:")
print(df.columns.tolist())

# 3. Identify possible webpage identifiers
page_columns = [
    col for col in df.columns
    if any(word in col.lower()
           for word in ["page", "url", "content_id", "page_id"])
]

print("\nPOSSIBLE WEBPAGE IDENTIFIERS:")
print(
    page_columns if page_columns
    else "No obvious webpage identifier found."
)

# 4. Define the decision supported by this dataset
decision = "Prioritise webpages for content review"
action_owner = "Content / SEO team"

wrong_call_costs = [
    "Wasted time and resources on low-priority webpages",
    "Important webpages needing attention may be overlooked",
    "Missed opportunities to improve content performance"
]

# 5. Display decision framework
print("\nDECISION FRAMEWORK")
print("-" * 50)
print("Decision:", decision)
print("Action owner:", action_owner)

print("\nCOST OF A WRONG RECOMMENDATION:")
for number, cost in enumerate(wrong_call_costs, 1):
    print(f"{number}. {cost}")

# 6. Display actual dataset records
print("\nSAMPLE DATA USED TO UNDERSTAND THE DECISION:")
display(df.head(10))
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 webpages and 44 columns. The average impressions over 90 days are 5,200.37, while the average sessions are 37.07. The average content age is 256.17 days. These numbers show that the dataset contains search performance and content-age information that can help identify webpages for further review. This makes the Refresh / Content Opportunity Scoring lane worth exploring over the next seven weeks.

```python
# Section 3: Quick Look at the FlyRank Dataset
# Direct connection to Hugging Face

!pip install -q datasets huggingface_hub

import pandas as pd
from datasets import load_dataset
from huggingface_hub import notebook_login

# --------------------------------------------------
# 1. Authenticate with Hugging Face
# --------------------------------------------------

notebook_login()

# --------------------------------------------------
# 2. Load the actual FlyRank dataset
# --------------------------------------------------

ds = load_dataset(
    "FlyRank/internship-warehouse",
    "fact_content_daily_performance",
    streaming=True,
    split="train"
)

print("Successfully connected to Hugging Face!")

# --------------------------------------------------
# 3. Inspect dataset features
# --------------------------------------------------

print("\nDATASET FEATURES")
print("-" * 50)
print(ds.features)

# --------------------------------------------------
# 4. Read 30,000 records
# --------------------------------------------------

records = list(ds.take(30000))
df = pd.DataFrame(records)

if df.empty:
    raise ValueError("The dataset returned no records.")

# --------------------------------------------------
# 5. Dataset overview
# --------------------------------------------------

print("\nDATASET OVERVIEW")
print("-" * 50)
print("Source: Hugging Face")
print("Records:", len(df))
print("Columns:", len(df.columns))

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

# --------------------------------------------------
# 6. Calculate numerical statistics
# --------------------------------------------------

numeric_df = df.select_dtypes(include="number")

print("\nNUMERICAL SUMMARY")
display(numeric_df.describe().T.round(2))

# --------------------------------------------------
# 7. Select 3 meaningful statistics
# --------------------------------------------------

# Map to columns actually present in the Hugging Face dataset
required_columns = [
    "gsc_impressions",
    "ga4_sessions",
    "report_date"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Expected columns not found: {missing_columns}"
    )

# Convert report_date to datetime to compute span metrics
df["report_date"] = pd.to_datetime(df["report_date"])
latest_date = df["report_date"].max()
df["days_since_report"] = (latest_date - df["report_date"]).dt.days

statistics = pd.DataFrame({
    "Metric": [
        "Average GSC Impressions",
        "Average GA4 Sessions",
        "Average days since report date"
    ],
    "Value": [
        df["gsc_impressions"].mean(),
        df["ga4_sessions"].mean(),
        df["days_since_report"].mean()
    ]
})

statistics["Value"] = statistics["Value"].round(2)

print("\nTHREE KEY DATASET FINDINGS")
display(statistics)

# --------------------------------------------------
# 8. Check missing values
# --------------------------------------------------

print("\nMISSING VALUE ANALYSIS")
display(
    (df.isnull().mean() * 100)
    .round(2)
    .sort_values(ascending=False)
    .to_frame("Missing Percentage")
)

# --------------------------------------------------
# 9. Final interpretation
# --------------------------------------------------

print("\nWHY THIS LANE IS WORTH EXPLORING")
print(
    "The dataset contains webpage search performance, "
    "traffic and content-age information. "
    "These features can support the prioritisation "
    "of webpages for content review."
)
```

## 4. Careful words: what I can and can't claim
